```python
# Lesson 5 - Functions 

# Basic function structure
def kelvin_to_celsius(kelvin):
    """
    convert temperature from Kelvin to Celsius.
    Landsat 8 thermal band gives values in Kelvin.
    """
    celsius = kelvin - 273.15
    return celsius

# Call the function 
result = kelvin_to_celsius(321.35)
print(f"Temperature: {result:.1f}°C")

# Call it multiple times with different inputs
print(f"{kelvin_to_celsius(319.95):.1f}°C")
print(f"{kelvin_to_celsius(317.25):.1f}°C")
print(f"{kelvin_to_celsius(314.45):.1f}°C")
print(f"{kelvin_to_celsius(311.85):.1f}°C")



```

```python
# Function with multiple parameters
def compute_ndvi(nir, red):
    """
    Compute Normalised Difference Vegetation Index.
    NDVI = (NIR - Red) / (NIR + Red)
    Range: -1 to 1. Higher = more vegetation.
    """
    ndvi = (nir - red) / (nir + red)
    return ndvi

# Test it 
print(f"NDVI: {compute_ndvi(0.45, 0.08):.3f}")
print(f"NDVI: {compute_ndvi(0.12, 0.09):.3f}")

# Function with default parameter values
def classify_zone(lst, critical=47, high=44, medium=41):
    """
    Classify a zone by heat risk level based on LST.
    Thresholds can be customized per city.
    """
    if lst >= critical:
         return "CRITICAL"
    elif lst >= high:
        return "HIGH"
    elif lst >= medium:
        return "MEDIUM"
    else:
        return "LOW"

# Using default thresholds
print(classify_zone(48.2))
print(classify_zone(44.1))
print(classify_zone(41.3))
print(classify_zone(38.7))

# Override thresholds for a cooler city
print("--- Guwahati thresholds ---")
print(classify_zone(38.7, critical=40, high=37, medium=34))
        
```

```python
# Functions calling other functions

def process_zone(zone_name, lst_kelvin, nir, red):
    """
    Takes raw satellite values for a zone and returns
    a complete processed dictionary ready for analysis.
    """
    # Call our existing functions inside this function
    lst = kelvin_to_celsius(lst_kelvin)
    ndvi = compute_ndvi(nir, red)
    risk = classify_zone(lst)
    
    # Return a complete zone dictionary
    return {
        "name": zone_name,
        "lst_celsius": round(lst, 1),
        "ndvi": round(ndvi, 3),
        "risk": risk
    }

# Process a single zone
walled_city = process_zone("Walled City", 321.35, 0.12, 0.09)
print(walled_city)

# Process all zones at once using a loop
raw_zones = [
    ("Walled City",   321.35, 0.12, 0.09),
    ("Johari Bazaar", 319.95, 0.14, 0.11),
    ("Sindhi Camp",   317.25, 0.18, 0.13),
    ("Civil Lines",   314.45, 0.28, 0.16),
    ("Mansarovar",    311.85, 0.35, 0.14),
]

processed = []
for name, kelvin, nir, red in raw_zones:
    zone = process_zone(name, kelvin, nir, red)
    processed.append(zone)

# Print results neatly
print("\n" + "=" * 50)
print("TAPASCAN — JAIPUR ZONE REPORT")
print("=" * 50)
for z in processed:
    print(f"{z['name']:<20} {z['lst_celsius']}°C  NDVI:{z['ndvi']}  [{z['risk']}]")

# Summary stats
lsts = [z["lst_celsius"] for z in processed]
print("=" * 50)
print(f"Peak LST:    {max(lsts)}°C")
print(f"Average LST: {sum(lsts)/len(lsts):.1f}°C")
print(f"Coolest:     {min(lsts)}°C")
```

```python
def summarize_city(zones, city_name):
    
    # 1. Count critical zones
    critical_count = 0
    for zone in zones:
        if zone["risk"] == "CRITICAL":
            critical_count += 1
    
    # 2. Average LST
    lsts = [z["lst_celsius"] for z in zones]
    avg = sum(lsts) / len(lsts)
    
    # 3. Hottest zone
    hottest = max(zones, key=lambda z: z["lst_celsius"])
    
    # 4. Print summary
    print(f"City: {city_name}")
    print(f"Critical zones: {critical_count}")
    print(f"Average LST: {avg:.1f}°C")
    print(f"Hottest zone: {hottest['name']}")

# Test it
summarize_city(processed, "Jaipur")
```
