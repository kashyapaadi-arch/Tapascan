```python
# Lesson 2 - Arithmetic and f-strings

# Temperature conversion - this is real Tapascan math
lst_kelvin = 321.35
lst_celsius = lst_kelvin - 273.15
print(lst_celsius)

# All arithmetic operators
a = 10
b = 3
print(a + b)
print(a - b)
print(a * b)
print(a / b)
print(a // b)
print(a % b)
print(a ** b)

# f-strings - learn this format, use it always
city = "Jaipur"
lst = 48.2
print(f"Peak LST in {city}: {lst}°C")

# f-string with rounding - the :.1f part 
lst_kelvin = 321.35
print(f"In Celsius: {lst_kelvin - 273.15:.1f}°C")
```

```python
# Lesson 3 - Lists

# A list of heat zones
zones = ["Walled City", "Johari Bazaar", "Sindhi Camp", "Civil Lines", "Mansarovar"]

#Access bhy index - starts at 0
print(zones[0])
print(zones[1])
print(zones[-1]) #last

#Length 
print(len(zones))

# Add and remove
zones.append("Vaishali Nagar")
print(len(zones))

zones.remove("Sindhi Camp")
print(zones)

# Check membership
print("Civil Lines" in zones)
print("Jaipur" in zones)

# Slicing
print(zones[0:3])
print(zones[1:])

# List of temperatures 
lst_values = [48.2, 46.8, 44.1, 41.3, 38.7]

print(max(lst_values))
print(min(lst_values))
print(sum(lst_values))
print(sum(lst_values) / len(lst_values)) 
```

```python
# Lesson 4 - Dictionaries

# One Zone with all its data in one place 
zone = { 
    "name": "Walled City Core",
    "lst": 48.2,
    "risk": " CRITICAL",
    "impervious_pct": 94,
    "ndvi": 0.02,
    "population": 87000
}

# Access values by key
print(zone["name"])
print(zone["lst"])
print(zone["population"])

# Add a new key
zone["state"] = "Rajasthan"
print(zone["state"])

# Check if key exists
print("risk" in zone)
print("albedo" in zone)

# Safe access — won't crash if key missing
print(zone.get("albedo", "not measured"))
print(zone.get("lst", "not measured"))

# See all keys and values
print(zone.keys())
print(zone.values())

# List of dictionaries — core Tapascan data structure
all_zones = [
    {"name": "Walled City",   "lst": 48.2, "risk": "CRITICAL"},
    {"name": "Johari Bazaar", "lst": 46.8, "risk": "CRITICAL"},
    {"name": "Civil Lines",   "lst": 41.3, "risk": "MEDIUM"},
    {"name": "Mansarovar",    "lst": 38.7, "risk": "LOW"},
]

# Access second zone's LST
print(all_zones[1]["lst"])

# Access last zone's risk
print(all_zones[-1]["risk"])

# Loop over all zones and print neatly
for z in all_zones:
    print(f"{z['name']} — {z['lst']}°C — {z['risk']}")
```
