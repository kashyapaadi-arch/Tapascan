```python
# Module 2 — Pandas
import pandas as pd
import numpy as np

print(pd.__version__)
```

```python
# Lesson 1 — Creating DataFrames

# From a dictionary — most common way
data = {
    "zone":       ["Walled City", "Johari Bazaar", "Sindhi Camp", 
                   "Civil Lines", "Mansarovar"],
    "lst":        [48.2, 46.8, 44.1, 41.3, 38.7],
    "ndvi":       [0.143, 0.120, 0.161, 0.273, 0.429],
    "risk":       ["CRITICAL", "HIGH", "HIGH", "MEDIUM", "LOW"],
    "population": [87000, 54000, 38000, 29000, 15000],
    "impervious": [94, 89, 76, 58, 42],
}

df = pd.DataFrame(data)

# View the dataframe
print(df)
print()

# Basic info
print(df.shape)      # rows, columns
print(df.dtypes)     # data type of each column
print(df.columns)    # column names
```

```python
# Lesson 2 — Accessing DataFrame data

# Access a single column — returns a Series
print(df["lst"])
print()

# Access multiple columns
print(df[["zone", "lst", "risk"]])
print()

# Access a single row by index
print(df.iloc[0])    # first row
print()

# Access by condition — boolean masking
hot = df[df["lst"] > 44]
print(hot)
print()

# Access specific value
print(df.iloc[0]["zone"])     # Walled City
print(df.iloc[0]["lst"])      # 48.2

# Quick statistics
print(df["lst"].mean())
print(df["lst"].max())
print(df["population"].sum())

# Full summary statistics
print(df.describe())
```

```python
# Lesson 3 — Adding columns and saving data

# Add a heat stress column
df["heat_stress"] = df["lst"] * (1 - df["ndvi"])
print(df[["zone", "lst", "ndvi", "heat_stress"]].round(2))
print()

# Add a cooling needed column
# How many degrees each zone needs to drop to reach safe threshold
df["cooling_needed"] = (df["lst"] - 40).clip(lower=0)
print(df[["zone", "lst", "cooling_needed"]])
print()

# Add intervention recommendation
def recommend(row):
    if row["risk"] == "CRITICAL":
        return "Immediate — tree corridors + cool roofs + water body"
    elif row["risk"] == "HIGH":
        return "Urgent — tree corridors + cool roofs"
    elif row["risk"] == "MEDIUM":
        return "Moderate — increase green cover"
    else:
        return "Monitor — maintain current vegetation"

df["recommendation"] = df.apply(recommend, axis=1)
print(df[["zone", "risk", "recommendation"]])
print()

# Sort by LST descending
df_sorted = df.sort_values("lst", ascending=False)
print(df_sorted[["zone", "lst", "risk"]])
print()

# Save to CSV
df.to_csv("jaipur_zones.csv", index=False)
print("Saved to jaipur_zones.csv")
```
